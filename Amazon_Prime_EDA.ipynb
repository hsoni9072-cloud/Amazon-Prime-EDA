## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Duplicate rows remove karein
titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)

# 2. Missing values handle karein
# Movies mein seasons NaN hota hai, isliye 0 fill karein
titles['seasons'] = titles['seasons'].fillna(0).astype(int)
titles['age_certification'] = titles['age_certification'].fillna('Not Rated')
titles['description'] = titles['description'].fillna('No Description')
credits['character'] = credits['character'].fillna('Unknown / Director')

# Numeric scores ke missing values ko median se fill karein
for col in ['imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']:
    titles[col] = titles[col].fillna(titles[col].median())

# 3. String lists ko actual Python lists mein convert karein (genres aur production_countries)
titles['genres_list'] = titles['genres'].apply(ast.literal_eval)
titles['countries_list'] = titles['production_countries'].apply(ast.literal_eval)

# 4. Exploded DataFrames banayein (Charts mein genres aur countries count karne ke liye)
exploded_genres = titles.explode('genres_list')
exploded_countries = titles.explode('countries_list')

# 5. Titles aur Credits ko merge karein (Actor/Director analysis ke liye)
merged_df = pd.merge(titles, credits, on='id', how='inner')

print("Data Wrangling Complete!")
print("Titles Shape:", titles.shape)
print("Merged Dataset Shape:", merged_df.shape)
print("Remaining Null Values in Titles:\n", titles.isnull().sum()[titles.isnull().sum() > 0])
````

### What all manipulations have you done and insights you found?

### Data Manipulations Performed:
1. **Duplicate Removal:** Checked and removed any duplicate rows across both `titles` and `credits` datasets to ensure accurate counts.
2. **Handling Missing Values:**
   * Filled missing `seasons` values with `0` (as missing values strictly represent `MOVIE` entries rather than `SHOW`).
   * Replaced missing `age_certification` values with `'Not Rated'` and missing `character` entries in `credits` with `'Unknown / Director'`.
   * Imputed missing numerical ratings and popularity metrics (`imdb_score`, `imdb_votes`, `tmdb_popularity`, `tmdb_score`) using their respective **medians** to prevent distortion from outliers.
3. **Parsing List Columns:** Converted string representations of lists in `genres` and `production_countries` into actual Python lists using `ast.literal_eval`, then created exploded views (`exploded_genres` and `exploded_countries`) for accurate category-level aggregation.
4. **Dataset Merging:** Performed an inner join between `titles` and `credits` on `id` (`merged_df`) to enable cast and director performance analysis.

### Key Wrangling Insights:
* Nearly all missing values in `seasons` belonged to Movies, confirming structural missingness rather than data loss.
* Many titles belong to multiple genres and countries simultaneously; exploding these columns is essential before plotting distributions.
